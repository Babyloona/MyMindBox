### Откуда брать задачу

Хорошая задача выглядит так: есть человек, который сейчас тратит на это время, и есть способ проверить, стало ли лучше.

Примеры, которые обычно получаются сильными:

- разбор обращений или заявок по категориям и приоритетам
- ассистент по внутренним документам компании или вуза
- помощник оператора: подсказывает ответ, человек подтверждает
- извлечение структурированных данных из писем, накладных, договоров
- проверка текстов на соответствие требованиям или регламенту

Примеры, которые обычно получаются слабыми: чат-бот общего назначения, «спроси что угодно по интернету», пересказ статей. У них нет измеримой задачи, и оценивать нечего.

## Правила игры

Это правила курса, на проекте они обязательны.

**1. Лестница вмешательства обязательна.** Занятие 45 начиналось с того, что способов повлиять на модель шесть, и дообучение среди них четвёртый, а не первый. В проекте вы обязаны сравнить **минимум три ступени** на одном и том же наборе:

- нулевая: базовый вызов модели без ухищрений
- первая: проработанный промпт, few-shot, структурированный вывод
- дальше ваша ступень по треку: retrieval, инструменты, дообучение, workflow

Работа, где сразу сделан агент или дообучение без сравнения с промптом, теряет баллы в двух критериях: и за методы адаптации, и за дисциплину замера.

**2. Оценочный набор фиксируется до начала работы.** Минимум тридцать примеров с эталонами. Собирается один раз, дальше не меняется. Все замеры на нём.

**3. Одно изменение за раз.** Поменяли промпт, модель и retrieval вместе, значит не узнали ничего.

**4. Числа сопровождаются неопределённостью.** На тридцати примерах разница в один ответ это три процента метрики. Занятие 46 было целиком про это: прежде чем объявлять улучшение, посчитайте интервал.

**5. Наблюдаемость не опциональна.** Каждый прогон должен быть виден в трекере: шаги, время, стоимость. Приложение без трейсинга это приложение, поведение которого вы не наблюдаете.

**6. Деньги считаются.** Стоимость одного обращения и стоимость обработки тысячи обращений. Это первый вопрос, который задаст заказчик.

**7. Ограничения называются вслух.** Работа, где всё получилось и ничего не сломалось, означает либо слишком простую задачу, либо неполный отчёт.

````python

RESULTS = []

def add_result(name, score, step="", split="eval", note="", secs=None,
               cost_usd=None, **extra):
    '''Записать вариант в лидерборд.

    name     : как называется вариант, например "ступень 1: few-shot"
    score    : значение основной метрики
    step     : ступень лестницы: "0", "1", "2", "4", "6" или своя пометка
    note     : что именно изменено по сравнению с предыдущим вариантом
    secs     : среднее время одного обращения
    cost_usd : средняя стоимость одного обращения
    '''
    RESULTS.append({"name": name, "score": float(score), "step": str(step),
                    "split": split, "note": note, "secs": secs,
                    "cost_usd": cost_usd, **extra})
    print(f"[лидерборд] {name}: {score:.3f}" +
          (f" | {secs:.2f} сек" if secs else "") +
          (f" | {cost_usd:.5f} USD" if cost_usd else ""))


def show_leaderboard(higher_is_better=True):
    if not RESULTS:
        print("лидерборд пуст")
        return None
    df = pd.DataFrame(RESULTS).sort_values("score", ascending=not higher_is_better)
    cols = [c for c in ["step", "name", "score", "secs", "cost_usd", "split", "note"]
            if c in df.columns]
    display(df[cols].reset_index(drop=True))

    fig, ax = plt.subplots(figsize=(9, 0.55 * len(df) + 1.2))
    ax.barh(df["name"], df["score"], color="#1C4A6E")
    for i, v in enumerate(df["score"]):
        ax.text(v, i, f" {v:.3f}", va="center", fontsize=10)
    ax.invert_yaxis()
    ax.set_xlabel("основная метрика")
    ax.set_title("Лидерборд проекта")
    ax.grid(True, axis="x", color="#D9DEE3", lw=0.8)
    ax.set_axisbelow(True)
    for s in ("top", "right", "left"):
        ax.spines[s].set_visible(False)
    plt.tight_layout(); plt.show()
    return df


def data_fingerprint(items, key="input"):
    '''Отпечаток оценочного набора: шестнадцать символов от отсортированных входов.

    Нужен, чтобы доказать, что все замеры сделаны на одном и том же наборе.
    Если отпечаток изменился между замерами, сравнивать их нельзя.
    '''
    joined = "\n".join(sorted(str(it[key]) for it in items))
    return hashlib.sha256(joined.encode("utf-8")).hexdigest()[:16]


def evaluate(items, predict_fn, scorers, name, step="", note="", show_errors=5):
    '''Прогнать вариант по оценочному набору.

    items      : список словарей с ключами input и expected
    predict_fn : функция, принимает строку вопроса, возвращает строку ответа
    scorers    : словарь {имя: функция(item, output) -> 0.0 или 1.0}
    '''
    rows, t0 = [], time.time()
    for it in items:
        started = time.time()
        try:
            out = predict_fn(it["input"])
        except Exception as exc:
            out = f"[ОШИБКА: {type(exc).__name__}: {str(exc)[:80]}]"
        row = {"input": it["input"], "expected": it.get("expected"), "output": out,
               "sec": round(time.time() - started, 2)}
        for sname, fn in scorers.items():
            try:
                row[sname] = float(fn(it, out))
            except Exception:
                row[sname] = 0.0
        rows.append(row)
    total = time.time() - t0

    df = pd.DataFrame(rows)
    main = list(scorers)[0]
    score = df[main].mean()
    n = len(df)
    se = math.sqrt(max(score * (1 - score), 1e-9) / n)

    print(f"=== {name} ===")
    print(f"примеров: {n} | время всего: {total:.0f} сек | среднее на пример: {df['sec'].mean():.2f} сек")
    for sname in scorers:
        print(f"  {sname}: {df[sname].mean():.3f}")
    print(f"интервал 95 процентов для основной метрики: плюс-минус {1.96 * se:.3f}")
    print(f"шаг метрики (один пример): {1 / n:.3f}")

    if show_errors:
        bad = df[df[main] < 1.0].head(show_errors)
        if len(bad):
            print(f"\nпримеры ошибок ({len(df[df[main] < 1.0])} всего):")
            for _, r in bad.iterrows():
                print(f"  вопрос : {str(r['input'])[:90]}")
                print(f"  эталон : {str(r['expected'])[:90]}")
                print(f"  ответ  : {str(r['output'])[:120]}")
                print()

    add_result(name, score, step=step, note=note, secs=df["sec"].mean())
    return df


def compare_runs(df_a, df_b, metric, name_a="вариант A", name_b="вариант B"):
    '''Парное сравнение двух вариантов на одном наборе.

    Сравнивать два средних недостаточно: информация о разнице живёт
    в примерах, где варианты разошлись. Это критерий Макнемара.
    '''
    from scipy.stats import binomtest
    a, b = df_a[metric].values, df_b[metric].values
    assert len(a) == len(b), "варианты мерились на разном числе примеров, сравнение незаконно"

    both_right = int(((a == 1) & (b == 1)).sum())
    both_wrong = int(((a == 0) & (b == 0)).sum())
    only_a = int(((a == 1) & (b == 0)).sum())
    only_b = int(((a == 0) & (b == 1)).sum())

    print(f"{name_a} против {name_b}")
    print(f"  обе правы          : {both_right}")
    print(f"  обе неправы        : {both_wrong}")
    print(f"  прав только {name_a}: {only_a}")
    print(f"  прав только {name_b}: {only_b}")
    print(f"  информативных примеров: {only_a + only_b} из {len(a)}")

    if only_a + only_b == 0:
        print("  вердикт: варианты совпали на всех примерах, разницы нет")
        return 1.0
    p = min(binomtest(min(only_a, only_b), only_a + only_b, 0.5).pvalue * 2, 1.0)
    print(f"  вероятность случайности: {p:.3f}")
    print("  вердикт:", "различие устойчиво" if p < 0.05 else "перевес неотличим от монеты")
    return p


def cost_report(prompt_tokens, completion_tokens, price_in, price_out,
                requests_per_month=1000, usd_kzt=458.87):
    '''Стоимость одного обращения и месячная стоимость. Цены смотрите у провайдера.'''
    per_req = prompt_tokens / 1e6 * price_in + completion_tokens / 1e6 * price_out
    print(f"токенов на обращение: {prompt_tokens} вход, {completion_tokens} выход")
    print(f"одно обращение : {per_req:.5f} USD = {per_req * usd_kzt:.2f} тенге")
    print(f"{requests_per_month} обращений: {per_req * requests_per_month:.2f} USD = "
          f"{per_req * requests_per_month * usd_kzt:.0f} тенге")
    return per_req

print("готово: add_result, show_leaderboard, data_fingerprint, evaluate, compare_runs, cost_report")
````

# 2. Бизнес-задача и пользователь

**1. Кто пользователь и что он делает сейчас?**

Человек, у которого не держатся в голове бытовые детали: где лежат провода, в какой папке паспорт, какой документ нужен для акта, что хотелось купить. Сейчас он пишет такое себе в чат, в «Избранное» или в заметки, а потом вручную листает, вспоминает точные слова поиска или не находит вовсе. Первый пользователь проекта сам автор, дальше ближайшие знакомые.

**2. Что именно делает система и где она встраивается в его работу?**

Бот в Telegram. Пользователь пишет факт или присылает фото и файл с подписью (запись), а потом спрашивает своими словами («где зарядка?», «скинь шаблон акта», «что мне надо сделать») и получает саму заметку или файл. Бот не заменяет человека: он достаёт записанное, а решение что с ним делать остаётся за пользователем. При неуверенности (ветка «не уверен, возможно, это оно») бот отдаёт находку с оговоркой.

**3. Какой измеримый эффект вы обещаете?**

Время поиска вручную порядка одной-трёх минут в лучшем случае но обычно поиск может занимать больше 10 мин (это оценка лично моя, а не замер, и в отчёте названа оценкой). Ответ бота занимает секунды: медиана 1.56 с и среднее 1.95 с по журналу из 156 запросов, в среднем 1.73 вызова LLM на запрос. Качество замерено на оценочном наборе из 49 примеров в разделе 10, там же стоимость одного обращения.

**4. Что будет, если система ошибётся?**

Цена ошибки небольшая, но есть два разных случая. Если бот не нашёл записанное, пользователь потратит пару минут на ручной поиск внутри чата, как и было до этого. Если бот выдал не ту заметку уверенно, пользователь может ей поверить, и это хуже. Поэтому система устроена так, чтобы ошибаться в сторону «не нашёл»: порог, LLM-проверка и честный отказ. Пароли и номера документов лучше в бот не записывать (предупреждение показано в `/start`). Да и в худшем случае пользователь прочтет и поймет что бот ошибся так как сам пользователь вносил данные.

**5. Почему для этого нужна LLM?**

Поиск по точному совпадению слов не находит «где зарядка?», когда записано «провода в чёрном чемодане». Эту часть решают эмбеддинги, а LLM здесь не нужна. LLM добавлена там, где эмбеддинги бессильны: они измеряют, о чём похоже, а не отвечает ли заметка на вопрос. Заметка «ключи от дачи» очень похожа на вопрос «где ключи от гаража?», и порог их не разделит. Это проверено на данных в разделе 8, а полезность LLM-проверки измерена парным сравнением ступеней 3 и 4. Я llm использовала как страховка и для классифицировании.

# 3. Архитектура и план реализации

**Поток данных.** Пользователь пишет в Telegram, бот (aiogram) передаёт сообщение в граф LangGraph, граф работает с векторной базой Chroma и вызывает LLM только в двух местах. Ответ и файл (по `file_id`) возвращаются в Telegram. Каждый запрос пишется в журнал и, если заданы ключи, уходит в трекер Langfuse.

```
сообщение ─ есть файл? ─ да ─────────────────────────────► save ─► конец
                │ нет
                ▼
            classify (LLM) ─ save ───────────────────────► save
                │ search / search_all
                ▼
            retrieve (эмбеддинги + Chroma, фильтр по user_id)
                ▼
            grade (порог без LLM) ─ ниже порога ─────────► «такого не записано»
                │ есть кандидаты
                ▼
            verify (LLM: отвечает ли заметка на вопрос) ─ нет ─► «такого не записано»
                │ да
                ├─ балл выше порога ─► уверенный ответ (заметка и файл)
                └─ балл неуверенный ─► ответ с оговоркой («возможно, это оно»)
```

**Где человек в контуре.** Бот ничего не удаляет и не принимает решений за человека: он достаёт записанное. В неуверенном случае оговорка прямо в тексте, и пользователь сам решает, то ли это. Фото без подписи бот не принимает и просит подпись, потому что поиск идёт только по тексту.

**Альтернативы, которые рассмотрели и отклонили:**

| решение | что выбрали | что отклонили и почему |
|---|---|---|
| векторная база | Chroma, файл на диске | pgvector и отдельный сервер: для сотен или тысяч заметок избыточно, нужна лишняя инфраструктура |
| эмбеддинги | локальная multilingual-MiniLM | OpenAI-эмбеддинги: качество в пределах шума (сравнение в разделе 9), зато локальные бесплатны и не зависят от платного ключа |
| только порог без LLM | порог как дешёвый предфильтр, LLM проверяет кандидата | чистый порог: эмбеддинги путают близкие по теме заметки, это измерено на ступенях 2 и 4 |
| оркестрация | LangGraph | цепочка `if`: работала бы, но граф даёт наглядную схему, единое состояние и явные ветки отказа и неуверенности |
| дообучение | не используем | своих данных для обучения нет, готовые модели справляются, а ступень 4 дешёвая |
| трекер | Langfuse Cloud | собственный Langfuse в Docker: шесть контейнеров и около 9 ГБ памяти, для проекта избыточно |
| Telegram | aiogram | python-telegram-bot: оба подходят, aiogram асинхронный и проще совмещается с health-сервером |

# 5. Метрика и оценщики

Что именно вы считаете правильным ответом. Это нужно решить **до** первого замера, иначе метрика подстроится под результат.

Два типа оценщиков, нужны оба:

- **правило**: проверяет проверяемое. Есть ли в ответе нужный идентификатор, правильная ли категория, соблюдён ли формат, есть ли ссылка на источник
- **судья**: другая модель оценивает то, что правилом не выразить. Не выдумала ли система данные, ответила ли на языке вопроса, отказалась ли там, где надо было отказаться

Про судью помните: он ошибается и склонен одобрять ответы, похожие на свои. Используйте его как дешёвый сигнал на объёме, а не как истину. Хорошая практика: разметить руками десять примеров и посмотреть, насколько судья с вами совпал.

````python
REFUSAL_PHRASES = ["не записано", "нет информации", "нет доступа", "не знаю",
                   "не нашел", "не нашла", "не могу найти", "нет такой", "нет такого"]

def norm(s):
    # Сравнение без учёта регистра, буквы ё и лишних пробелов: «чёрном» и «черном» одно и то же
    return " ".join(str(s).lower().replace("ё", "е").split())

def notes_in(output):
    low = norm(output)
    return sorted(norm(t) for t in ALL_NOTE_TEXTS if norm(t) in low)

def describe_expected(item):
    e = item["expected"]
    if e == "отказ":
        return "в личной базе такого нет: правильный ответ это отказ (такого не записано), без выдуманных заметок"
    if isinstance(e, list):
        return "должны быть выданы ровно эти заметки и никаких других: " + "; ".join(e)
    return "должна быть выдана ровно эта заметка и никакие другие: " + e


# Оценщик-правило. 1.0 или 0.0. Проверяет, какие заметки из базы встречаются в ответе.
# Правило зафиксировано до первого замера и между замерами не менялось.
def rule_score(item, output):
    out = norm(output)
    found = notes_in(output)
    expected = item["expected"]
    if expected == "отказ":
        has_refusal = any(p in out for p in REFUSAL_PHRASES)
        return 1.0 if (len(found) == 0 and has_refusal) else 0.0
    wanted = expected if isinstance(expected, list) else [expected]
    return 1.0 if found == sorted(norm(w) for w in wanted) else 0.0


# Быстрая самопроверка правила на трёх очевидных случаях
assert rule_score({"expected": "провода в чёрном чемодане на балконе"}, "Провода в черном чемодане на балконе.") == 1.0
assert rule_score({"expected": "отказ"}, "Такого не записано.") == 1.0
assert rule_score({"expected": "отказ"}, "Возможно: провода в чёрном чемодане на балконе") == 0.0
print("правило проверено на трёх случаях")


# Оценщик-судья на модели. Просим одну цифру, а не рассуждение.
JUDGE_PROMPT = """Ты оцениваешь ответ помощника личной памяти. Ответь ровно одной цифрой.
1 если ответ корректен по сути, не содержит выдуманных данных и соответствует ожиданию.
0 если есть выдумка, ответ не по существу или противоречит ожиданию.

ВОПРОС: {q}
ОЖИДАНИЕ: {e}
ОТВЕТ ПОМОЩНИКА: {o}

Только цифра."""

def judge_score(item, output):
    r = oa.chat.completions.create(
        model=MODEL, temperature=0,
        messages=[{"role": "user", "content": JUDGE_PROMPT.format(
            q=item["input"], e=describe_expected(item), o=output)}],
    )
    raw = (r.choices[0].message.content or "").strip()
    return 1.0 if raw.startswith("1") else 0.0


SCORERS = {"правило": rule_score, "судья": judge_score}
print("оценщиков:", list(SCORERS))
````

# 6. Ступень 0: базовый вызов

Первый замер. Модель без ухищрений: короткий вопрос, никакого контекста, никаких инструментов.

Смысл этой ступени тот же, что у бейзлайна в проекте по Deep Learning: **показать, сколько даёт задача сама по себе**. Дальше все проценты считаются относительно этого числа.

Часто выясняется, что модель и так знает половину ответов, и тогда ваша работа должна объяснить, зачем нужны остальные ступени.

````python
from app.llm import USAGE      # общий счётчик токенов: им же считает граф MindBox
TOKENS = {}

def oa_chat(messages):
    # Вызов модели напрямую, с учётом токенов в общем счётчике
    r = oa.chat.completions.create(model=MODEL, temperature=0, messages=messages)
    USAGE.calls += 1
    USAGE.prompt_tokens += r.usage.prompt_tokens
    USAGE.completion_tokens += r.usage.completion_tokens
    return r.choices[0].message.content


def run_step(predict_fn, name, step, note):
    # Прогон варианта по набору плюс токены и стоимость одного обращения
    USAGE.reset()
    df = evaluate(EVAL_SET, predict_fn, SCORERS, name=name, step=step, note=note)
    snap = USAGE.snapshot()
    count = len(EVAL_SET)
    tin, tout = snap["prompt_tokens"] / count, snap["completion_tokens"] / count
    cost = tin / 1e6 * PRICE_IN + tout / 1e6 * PRICE_OUT
    RESULTS[-1]["cost_usd"] = cost
    TOKENS[name] = {"calls": snap["calls"] / count, "in": tin, "out": tout, "cost": cost}
    print(f"токены на обращение: вход {tin:.0f}, выход {tout:.0f} | вызовов LLM: {snap['calls'] / count:.2f} | {cost:.5f} USD")
    return df


NAME0 = "ступень 0: базовый вызов"

def predict_step0(question):
    return oa_chat([{"role": "user", "content": question}])


df0 = run_step(predict_step0, NAME0, "0", "без промпта и без заметок")
````

# 8. Ваша ступень: поиск по смыслу и граф

Основная часть проекта разбита на три отдельных замера, чтобы менять по одной вещи.

**Ступень 2: эмбеддинги и порог, без LLM.** Каждая заметка превращается в вектор (локальная multilingual-MiniLM) и хранится в Chroma, вопрос превращается в вектор, берётся ближайшая заметка, и если сходство не ниже порога 0.40, она выдаётся, иначе отказ. Вопрос «покажи все» определяется по ключевому слову («все», «всё», «список»). Гипотеза: поиск по смыслу найдёт заметки, у которых нет общих слов с вопросом, и ступень 2 обгонит ступень 1 при росте базы. Ожидаем, что порог не отличит близкую по теме заметку от нужной.

**Ступень 3: плюс LLM-классификатор намерения.** Вместо ключевого слова модель определяет, хочет ли пользователь одну заметку или все (и общие вопросы про категорию вроде «что мне надо сделать»). Гипотеза: улучшатся вопросы «покажи все» без слова «все».

**Ступень 4: плюс LLM-проверка совпадения (граф MindBox целиком).** Найденная заметка проходит проверку: отвечает ли она на вопрос. Предфильтр по порогу остаётся, ступени «не уверен» и «не записано» добавлены. Гипотеза: исчезнут уверенные выдачи не той заметки («ключи от гаража» при заметке про ключи от дачи), потому что эмбеддинги измеряют тематическую близость, а не ответ на вопрос. Платой будет второй вызов LLM на запрос.

Параметры: порог 0.40 для уверенного ответа, 0.30 для «возможно, это оно», 0.20 для предфильтра вопросов «покажи все» (проверка LLM отсекает лишнее). Значения подобраны на наборе разработки и поэтому на нём оптимистичны, об этом сказано в разделе про данные.

````python
# Ступени 2 и 3 собраны из тех же функций базы, что использует бот
from app.graph import classify

NOT_FOUND = "Такого не записано."

def wants_all(question):
    q = " " + question.lower() + " "
    return any(w in q for w in [" все ", " всё ", " всех ", "список"])

def answer_by_threshold(question, all_mode):
    found = search_notes(EVAL_USER, question, top_k=10)
    if all_mode:
        good = [r for r in found if r["score"] >= config.THRESHOLD_FOUND]
        if len(good) == 0:
            return NOT_FOUND
        good = sorted(good, key=lambda r: r["created"])
        return "\n".join(r["text"] for r in good)
    best = found[0]
    if best["score"] >= config.THRESHOLD_FOUND:
        return best["text"]
    return NOT_FOUND

def predict_step2(question):
    return answer_by_threshold(question, wants_all(question))

def predict_step3(question):
    intent = classify({"text": question})["intent"]
    return answer_by_threshold(question, intent == "search_all")

# Ступень 4: граф MindBox целиком (classify, retrieve, grade, verify, три варианта ответа)
from app.tracing import get_trace_config

def predict_graph(question):
    state = {"user_id": EVAL_USER, "text": question, "file_id": "", "file_type": "text", "steps": []}
    cfg = get_trace_config(EVAL_USER, session_id="notebook-eval", tags=["notebook", "eval"])
    if cfg is not None:
        out = graph.invoke(state, config=cfg)
    else:
        out = graph.invoke(state)
    return out["answer"]

print("функции готовы")
````

# 9. Улучшение: одно изменение

**Что меняем:** только модель эмбеддингов в ступени 2. Локальная multilingual-MiniLM заменяется на `text-embedding-3-small` от OpenAI. Всё остальное (ключевое слово для «покажи все», структура ответа) то же. Порог у другой модели другой, поэтому берётся значение 0.30, подобранное на наборе разработки для этой модели (так честнее, чем оставлять 0.40 от чужой модели).

**Гипотеза:** OpenAI-эмбеддинги лучше ранжируют разговорные формулировки, поэтому ступень 2 улучшится. Ожидаю небольшой эффект, в пределах шума на 49 примерах. Если разницы нет, это тоже результат: тогда локальные эмбеддинги выигрывают по цене и независимости.

````python
from langchain_openai import OpenAIEmbeddings

OPENAI_THRESHOLD = 0.30
emb = OpenAIEmbeddings(model=os.getenv("OPENAI_EMBEDDING_MODEL", "text-embedding-3-small"))

note_keys = list(NOTES.keys())
note_vecs = np.array(emb.embed_documents([NOTES[k] for k in note_keys]))
question_vecs = {}
vecs = emb.embed_documents([it["input"] for it in EVAL_SET])
for it, v in zip(EVAL_SET, vecs):
    question_vecs[it["input"]] = np.array(v)

def cosine_scores(qv):
    return note_vecs @ qv / (np.linalg.norm(note_vecs, axis=1) * np.linalg.norm(qv))

def predict_improved(question):
    scores = cosine_scores(question_vecs[question])
    order = np.argsort(-scores)
    if wants_all(question):
        good = [NOTES[note_keys[i]] for i in order[:10] if scores[i] >= OPENAI_THRESHOLD]
        return "\n".join(good) if good else NOT_FOUND
    best = order[0]
    if scores[best] >= OPENAI_THRESHOLD:
        return NOTES[note_keys[best]]
    return NOT_FOUND

NAME_IMP = "улучшение: OpenAI-эмбеддинги вместо локальных (ступень 2)"
df_imp = evaluate(EVAL_SET, predict_improved, SCORERS, name=NAME_IMP, step="2+",
                  note="изменено ровно одно: модель эмбеддингов, порог 0.30 для неё")
````

**Откуда гипотеза.** Из разбора ошибки по трейсу (раздел 11): на вопрос про шарлотку поиск первым ставит рецепт борща (оба про готовку), `grade` пропускает к проверке только его, `verify` правильно его отклоняет, и нужная заметка до проверки не доходит. Ожидаем, что вырастут разговорные вопросы, где нужная заметка вторая или третья по сходству. Риск: с тремя кандидатами LLM чаще выберет что-то на вопрос без ответа, то есть могут просесть отказы.
````python
# Прогон выбранных вопросов с трейсингом. Метаданные: сессия, пользователь, теги, имя трейса.
# На каждый трейс ставим две оценки: правило и судья.
from langfuse.langchain import CallbackHandler

TRACE_SESSION = "notebook-demo-" + time.strftime("%Y%m%d-%H%M")
TRACE_QUESTIONS = [
    "где зарядка?",                       # типовой
    "скинь все акты",                     # общий
    "какой у меня пароль от wifi?",       # отказ
    "где ключи от гаража?",               # пограничный отказ
    "как приготовить шарлотку",           # разговорный
    "что я писала на счет посылки",       # разговорный
]
by_input = {it["input"]: it for it in EVAL_SET}
traced = []

if not ok:
    print("трекер недоступен, ячейка пропущена")
else:
    for q in TRACE_QUESTIONS:
        item = by_input[q]
        handler = CallbackHandler()
        cfg = {
            "callbacks": [handler],
            "run_name": "mindbox_notebook",
            "metadata": {
                "langfuse_session_id": TRACE_SESSION,
                "langfuse_user_id": "eval_user",
                "langfuse_tags": ["notebook", "demo", item["kind"]],
            },
        }
        state = {"user_id": EVAL_USER, "text": q, "file_id": "", "file_type": "text", "steps": []}
        out = graph.invoke(state, config=cfg)
        r_score = rule_score(item, out["answer"])
        j_score = judge_score(item, out["answer"])
        trace_id = getattr(handler, "last_trace_id", None)
        if trace_id:
            for sname, val in [("правило", r_score), ("судья", j_score)]:
                try:
                    lf.create_score(name=sname, value=val, trace_id=trace_id, data_type="NUMERIC")
                except Exception as e:
                    print("оценка не записалась:", type(e).__name__, str(e)[:80])
        traced.append({"вопрос": q, "правило": r_score, "судья": j_score,
                       "trace_id": trace_id, "путь": " -> ".join(out["steps"])})
    lf.flush()
    print("сессия в Langfuse:", TRACE_SESSION)
    for t in traced:
        mark = "верно " if t["правило"] == 1 else "ОШИБКА"
        print(f"{mark} | {t['вопрос']} | судья {t['судья']:.0f} | trace {t['trace_id']}")
        print(f"       путь: {t['путь']}")
    
    # До и после: тот же вопрос с проверкой трёх кандидатов, отдельный трейс
    config.VERIFY_TOP_K = 3
    handler = CallbackHandler()
    q = "как приготовить шарлотку"
    out = graph.invoke(
        {"user_id": EVAL_USER, "text": q, "file_id": "", "file_type": "text", "steps": []},
        config={"callbacks": [handler], "run_name": "mindbox_notebook_after_fix",
                "metadata": {"langfuse_session_id": TRACE_SESSION, "langfuse_user_id": "eval_user",
                             "langfuse_tags": ["notebook", "after-fix"]}})
    config.VERIFY_TOP_K = 1
    lf.flush()
    print("\nпосле улучшения (VERIFY_TOP_K=3):", q, "->", out["answer"][:60])
    print("       путь:", " -> ".join(out["steps"]), "| trace", getattr(handler, "last_trace_id", None))
````

````python
def self_check():
    '''Проверка перед сдачей. Запустите последней ячейкой.'''
    ok, warn = [], []

    placeholders = ["Фамилия Имя", "A / B / C / D / E / F", "Одно предложение",
                    "Название приложения", "ВАШ КОД", "Кто конкретно"]
    empty = [k for k, v in PROJECT.items()
             if not str(v).strip() or any(p in str(v) for p in placeholders)]
    if empty:
        warn.append("В PROJECT не заполнены: " + ", ".join(empty))
    else:
        ok.append("PROJECT заполнен")

    n_eval = len(EVAL_SET) if "EVAL_SET" in globals() else 0
    if n_eval < 30:
        warn.append(f"В оценочном наборе {n_eval} примеров, нужно минимум 30")
    else:
        ok.append(f"Оценочный набор: {n_eval} примеров")
        kinds = {it.get("kind", "не указан") for it in EVAL_SET}
        if len(kinds) < 2:
            warn.append("В наборе не размечены типы примеров: нет пограничных случаев")
        else:
            ok.append(f"Типы примеров в наборе: {', '.join(sorted(kinds))}")
        refusals = sum(1 for it in EVAL_SET if str(it.get("expected", "")).lower() == "отказ")
        if refusals == 0:
            warn.append("В наборе нет случаев, где правильный ответ это отказ")
        else:
            ok.append(f"Случаев с отказом: {refusals}")

    if "EVAL_HASH" in globals() and EVAL_HASH:
        current = data_fingerprint(EVAL_SET)
        if current != EVAL_HASH:
            warn.append("Отпечаток набора изменился после первого замера: "
                        "сравнивать прогоны между собой нельзя")
        else:
            ok.append(f"Отпечаток набора не менялся: {EVAL_HASH}")
    else:
        warn.append("Отпечаток оценочного набора не зафиксирован")

    steps = {r.get("step") for r in RESULTS}
    if len(RESULTS) < 3:
        warn.append(f"В лидерборде {len(RESULTS)} записей, нужно минимум 3 ступени")
    else:
        ok.append(f"Записей в лидерборде: {len(RESULTS)}, ступеней: {len(steps)}")
    if "0" not in steps:
        warn.append("Нет ступени 0: базовый вызов без промпта обязателен")
    notes = [r for r in RESULTS if not str(r.get("note", "")).strip()]
    if notes:
        warn.append(f"У {len(notes)} записей лидерборда не заполнено поле note: "
                    "непонятно, что менялось между вариантами")

    if "lf" in globals():
        ok.append("Трекер подключён")
    else:
        warn.append("Трейсинг не подключён")

    if "REPO" in globals():
        bad = [k for k, v in REPO.items() if "ВАШ КОД" in str(v)]
        if bad:
            warn.append("В REPO не заполнены: " + ", ".join(bad))
        else:
            ok.append("Репозиторий описан")
    else:
        warn.append("Раздел про продакшен не заполнен")

    print("ГОТОВО:")
    for line in ok:
        print("  +", line)
    print()
    if warn:
        print("НЕ ХВАТАЕТ:")
        for line in warn:
            print("  -", line)
    else:
        print("Всё на месте. Проверьте ещё раз, что в файлах нет ключей, и сдавайте.")
    return len(warn) == 0


self_check()
````
